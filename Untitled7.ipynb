### Example of using `ipywidgets` for position selection

Below is an example of how you can replace the text input for 'position' with a dropdown widget. When you run this cell, you'll see a dropdown menu. Select an option and then click 'Submit'. The selected value will be stored in the `position` variable.

```python
# Initialize variables
speed = 0
control = 0
power = 0

# Create a dropdown widget for position
position_choices = ['Ytter', 'Anfallare', 'Mittfältare', 'Back', 'Försvarare', 'Målvakt', 'Annan']
position_widget = widgets.Dropdown(
    options=position_choices,
    value='Annan',
    description='Välj position:'
)

# Create a button to submit the choice
submit_button = widgets.Button(description="Submit")

# Display the widget and button
display(position_widget, submit_button)

# Define an event handler for the button click
def on_submit_button_clicked(b):
    global speed, control, power, position # Declare global to modify the variables
    position = position_widget.value.lower()

    with clear_output(wait=True):
        display(position_widget, submit_button) # Re-display widgets to keep them interactive
        print(f"Du valde position: {position_widget.value}")

        if position == "ytter" or position == "anfallare":
            speed = speed + 2
        elif position == "mittfältare":
            control = control + 2
        elif position == "back" or position == "försvarare":
            power = power + 1
        elif position == "målvakt":
            control = control + 1
        else:
            print("Jag känner inte igen positionen, den påverkar inte resultatet.")

        print(f"Current scores: Speed={speed}, Control={control}, Power={power}")

# Attach the event handler to the button
submit_button.on_click(on_submit_button_clicked)
```

## Interaktiv Fotbollssko-väljare (med widgets)

Denna del av anteckningsboken innehåller en interaktiv version av frågesporten där du använder knappar och rullgardinsmenyer istället för att skriva in svar.

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Globala variabler för frågesportens tillstånd
name = ""
speed = 0
control = 0
power = 0
current_question_index = 0 # Används för att hålla reda på vilken fråga som är aktiv

# --- Välkomstmeddelande och Namninmatning (kan fortfarande vara textinmatning för enkelhetens skull) ---
print("-------------------------------FOOTBALL BOOT SELECTOR------------------------------")
print("Hej och välkommen till en enkel och precis guide till den perfekta fotbollsskon för dig. Genom några enkla frågor och svar kommer du snabbt hitta skon för just dig!")
name = input("Låt oss börja med ditt namn: ")
print("Hej", name, "kul att du hittade hit")

# --- Funktion för att visa frågor dynamiskt ---
def display_quiz_question():
    global speed, control, power, current_question_index, name

    clear_output(wait=True) # Rensa tidigare utdata innan nästa fråga visas

    # --- Fråga 0: Position ---
    if current_question_index == 0:
        position_choices = ['Ytter', 'Anfallare', 'Mittfältare', 'Back', 'Försvarare', 'Målvakt', 'Annan']
        position_widget = widgets.Dropdown(
            options=position_choices,
            description='Välj position:'
        )
        submit_button = widgets.Button(description="Nästa fråga")

        output_area = widgets.Output() # För att fånga print-satser som är kopplade till denna widget

        def on_position_submit(b):
            global speed, control, power, current_question_index
            position_answer = position_widget.value.lower()

            with output_area:
                clear_output() # Rensa bara output_area
                print(f"Du valde position: {position_widget.value}")
                if position_answer == "ytter" or position_answer == "anfallare":
                    speed += 2
                elif position_answer == "mittfältare":
                    control += 2
                elif position_answer == "back" or position_answer == "försvarare":
                    power += 1
                elif position_answer == "målvakt":
                    control += 1
                else:
                    print("Jag känner inte igen positionen, den påverkar inte resultatet.")
                print(f"Aktuella poäng: Snabbhet={speed}, Kontroll={control}, Kraft={power}")

            current_question_index += 1
            display_quiz_question() # Visa nästa fråga

        submit_button.on_click(on_position_submit)
        display(widgets.VBox([position_widget, submit_button, output_area]))

    # --- Fråga 1: Hur viktigt är snabbhet för dig? ---
    elif current_question_index == 1:
        svar_choices = ['Mycket', 'Ganska', 'Lite']
        svar_widget = widgets.RadioButtons(
            options=svar_choices,
            description='1. Hur viktigt är snabbhet för dig?'
        )
        submit_button = widgets.Button(description="Nästa fråga")
        output_area = widgets.Output()

        def on_svar1_submit(b):
            global speed, current_question_index
            svar_answer = svar_widget.value.lower()
            with output_area:
                clear_output()
                print(f"Du svarade: {svar_widget.value}")
                if svar_answer == "mycket":
                    speed += 3
                elif svar_answer == "ganska":
                    speed += 2
                elif svar_answer == "lite":
                    speed += 1
                print(f"Aktuella poäng: Snabbhet={speed}, Kontroll={control}, Kraft={power}")
            current_question_index += 1
            display_quiz_question()

        submit_button.on_click(on_svar1_submit)
        display(widgets.VBox([svar_widget, submit_button, output_area]))

    # --- Fråga 2: Hur viktigt är bollkontroll för dig? ---
    elif current_question_index == 2:
        svar_choices = ['Mycket', 'Ganska', 'Lite']
        svar_widget = widgets.RadioButtons(
            options=svar_choices,
            description='2. Hur viktigt är bollkontroll för dig?'
        )
        submit_button = widgets.Button(description="Nästa fråga")
        output_area = widgets.Output()

        def on_svar2_submit(b):
            global control, current_question_index
            svar_answer = svar_widget.value.lower()
            with output_area:
                clear_output()
                print(f"Du svarade: {svar_widget.value}")
                if svar_answer == "mycket":
                    control += 3
                elif svar_answer == "ganska":
                    control += 2
                elif svar_answer == "lite":
                    control += 1
                print(f"Aktuella poäng: Snabbhet={speed}, Kontroll={control}, Kraft={power}")
            current_question_index += 1
            display_quiz_question()

        submit_button.on_click(on_svar2_submit)
        display(widgets.VBox([svar_widget, submit_button, output_area]))

    # --- Fråga 3: Hur viktigt är kraft i dina skott för dig? ---
    elif current_question_index == 3:
        svar_choices = ['Mycket', 'Ganska', 'Lite']
        svar_widget = widgets.RadioButtons(
            options=svar_choices,
            description='3. Hur viktigt är kraft i dina skott för dig?'
        )
        submit_button = widgets.Button(description="Nästa fråga")
        output_area = widgets.Output()

        def on_svar3_submit(b):
            global power, current_question_index
            svar_answer = svar_widget.value.lower()
            with output_area:
                clear_output()
                print(f"Du svarade: {svar_widget.value}")
                if svar_answer == "mycket":
                    power += 3
                elif svar_answer == "ganska":
                    power += 3 # Originalkoden hade 'power=power+3' för både 'ganska' och 'mycket'
                elif svar_answer == "lite":
                    power += 1
                print(f"Aktuella poäng: Snabbhet={speed}, Kontroll={control}, Kraft={power}")
            current_question_index += 1
            display_quiz_question()

        submit_button.on_click(on_svar3_submit)
        display(widgets.VBox([svar_widget, submit_button, output_area]))

    # --- Fråga 4: Gillar du att dribbla mycket? ---
    elif current_question_index == 4:
        svar_choices = ['Ja', 'Nej']
        svar_widget = widgets.RadioButtons(
            options=svar_choices,
            description='4. Gillar du att dribbla mycket?'
        )
        submit_button = widgets.Button(description="Nästa fråga")
        output_area = widgets.Output()

        def on_svar4_submit(b):
            global control, speed, power, current_question_index
            svar_answer = svar_widget.value.lower()
            with output_area:
                clear_output()
                print(f"Du svarade: {svar_widget.value}")
                if svar_answer == "ja":
                    control += 3
                elif svar_answer == "nej":
                    speed += 1
                    power -= 1
                print(f"Aktuella poäng: Snabbhet={speed}, Kontroll={control}, Kraft={power}")
            current_question_index += 1
            display_quiz_question()

        submit_button.on_click(on_svar4_submit)
        display(widgets.VBox([svar_widget, submit_button, output_area]))

    # --- Fråga 5: Löper du mycket? ---
    elif current_question_index == 5:
        svar_choices = ['Ja', 'Nej']
        svar_widget = widgets.RadioButtons(
            options=svar_choices,
            description='5. Löper du mycket?'
        )
        submit_button = widgets.Button(description="Nästa fråga")
        output_area = widgets.Output()

        def on_svar5_submit(b):
            global speed, control, power, current_question_index
            svar_answer = svar_widget.value.lower()
            with output_area:
                clear_output()
                print(f"Du svarade: {svar_widget.value}")
                if svar_answer == "ja":
                    speed += 3
                elif svar_answer == "nej":
                    # speed modifierades inte för 'Nej' i originalkoden, så speed += 0
                    control += 1
                    power += 1
                print(f"Aktuella poäng: Snabbhet={speed}, Kontroll={control}, Kraft={power}")
            current_question_index += 1
            display_quiz_question()

        submit_button.on_click(on_svar5_submit)
        display(widgets.VBox([svar_widget, submit_button, output_area]))

    # --- Fråga 6: Skjuter du långskott ofta? ---
    elif current_question_index == 6:
        svar_choices = ['Ja', 'Nej']
        svar_widget = widgets.RadioButtons(
            options=svar_choices,
            description='6. Skjuter du långskott ofta?' # Korrigerat numrering (original hade två Q7)
        )
        submit_button = widgets.Button(description="Nästa fråga")
        output_area = widgets.Output()

        def on_svar6_submit(b):
            global power, speed, control, current_question_index
            svar_answer = svar_widget.value.lower()
            with output_area:
                clear_output()
                print(f"Du svarade: {svar_widget.value}")
                if svar_answer == "ja":
                    power += 3
                elif svar_answer == "nej":
                    # power modifierades inte för 'Nej' i originalkoden
                    speed += 1
                    control += 1
                print(f"Aktuella poäng: Snabbhet={speed}, Kontroll={control}, Kraft={power}")
            current_question_index += 1
            display_quiz_question()

        submit_button.on_click(on_svar6_submit)
        display(widgets.VBox([svar_widget, submit_button, output_area]))

    # --- Fråga 7: Passar du mycket? ---
    elif current_question_index == 7:
        svar_choices = ['Ja', 'Nej']
        svar_widget = widgets.RadioButtons(
            options=svar_choices,
            description='7. Passar du mycket?'
        )
        submit_button = widgets.Button(description="Nästa fråga")
        output_area = widgets.Output()

        def on_svar7_submit(b):
            global control, speed, power, current_question_index
            svar_answer = svar_widget.value.lower()
            with output_area:
                clear_output()
                print(f"Du svarade: {svar_widget.value}")
                if svar_answer == "ja":
                    control += 2
                elif svar_answer == "nej":
                    speed += 1
                    power += 1
                print(f"Aktuella poäng: Snabbhet={speed}, Kontroll={control}, Kraft={power}")
            current_question_index += 1
            display_quiz_question()

        submit_button.on_click(on_svar7_submit)
        display(widgets.VBox([svar_widget, submit_button, output_area]))

    # --- Fråga 8: Vad föredrar du? (1, 2 eller 3) ---
    elif current_question_index == 8:
        val_choices = ['1. Lätt sko', '2. Bra bollkänsla', '3. Stabil och kraftfull sko']
        val_widget = widgets.RadioButtons(
            options=val_choices,
            description='8. Vad föredrar du?'
        )
        submit_button = widgets.Button(description="Visa resultat")
        output_area = widgets.Output()

        def on_val_submit(b):
            global speed, control, power, current_question_index
            val_answer = val_widget.value[0] # Hämta bara siffran '1', '2' eller '3'
            with output_area:
                clear_output()
                print(f"Du svarade: {val_widget.value}")
                if val_answer == "1":
                    speed += 3
                elif val_answer == "2":
                    control += 3
                elif val_answer == "3":
                    power += 3
                print(f"Aktuella poäng: Snabbhet={speed}, Kontroll={control}, Kraft={power}")

            current_question_index += 1
            display_quiz_question() # Går till slutresultatet

        submit_button.on_click(on_val_submit)
        display(widgets.VBox([val_widget, submit_button, output_area]))

    # --- Slutresultat ---
    else: # current_question_index > 8 (alla frågor besvarade)
        print("                                 DITT RESULTAT!")
        print("-------------------------------------------------------------------------------")
        print("Snabbhets-poäng", speed)
        print("Kontroll-poäng:", control)
        print("Kraft-poäng", power)

        # Justerade logiken för att hantera 'störst' korrekt
        if speed > control and speed > power:
            print("Du verkar centrera ditt spel kring din snabbhet och acceleration.")
            print("En smidig och lätt fotbollssko skulle nog därför passa din spelstil.")
            print("Rekommenderade skor:")
            print("Adidas F50 League, pris:839 kr")
            print("Nike mercurial vapor 16 elite, pris:2799 kr")
        elif control > speed and control > power:
            print("Du verkar värdesätta bollkontroll högt i din spelstil.")
            print("En mönstrad och vadderad topp rekommenderas på din sko.")
            print("Rekommenderade skor:")
            print("Nike Phantom GX 2 Elite, pris: 2939 kr")
            print("Adidas Predator Elite Fold-Over, pris:3299 kr")
        elif power > speed and power > control:
            print("Du verkar leta efter en kraftfull och robust sko.")
            print("Dessa skor är ganska allround och funkar till det allra mesta utan att kompensera andra områden alltför mycket.")
            print("Rekommenderade skor: ")
            print("Puma King Ultimate, pris 1225 kr")
            print("Puma King 9 Matchk, pris 1099 kr")
        else: # Hantera oavgjorda resultat eller balanserade poäng
            print("Det är svårt att avgöra en dominant egenskap baserat på dina svar, eller så har du en välbalanserad spelstil.")
            print("Du kan ha nytta av en allround-sko eller en modell som förstärker din favoritfunktion.")
            print("Rekommenderade skor för allround-spelare:")
            print("Adidas Copa Sense.1 FG, pris: 1999 kr")
            print("Nike Tiempo Legend 9 Elite FG, pris: 2199 kr")

        print("Tack för att du har delatagit i vårat quiz och hoppas du är nöjd med ditt resultat och hittar din perfekta sko.")
        print("På återseende!")

# Starta frågesporten
display_quiz_question()
```
